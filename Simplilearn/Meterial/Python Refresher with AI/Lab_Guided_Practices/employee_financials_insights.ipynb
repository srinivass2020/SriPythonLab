```python
print('Employee Finanaial Insights System')
# Collect user input
name = input("Enter emloyee name: ")
department = input("Enter employee department: ")
experience = int(input("Enter years of experience: "))
salary = float(input("Enter the salary ($): "))
annualSalary = salary*12;
taxDeduction=(30*salary)/100;
yearlyIncrement = 5/100;
yearlySavings=25/100;

print(f"Employee Name : ",name)
print(f"Employee Department : ",department)
print(f"Employee Experience : ",experience)
print(f"Employee Monthly Salary : ",salary)
print(f"Employee Annual Salary : ",annualSalary)
print(f"Employee Monthly Tax : ",taxDeduction)
print(f"Employee Monthly NetIncome : ",salary-taxDeduction)

y1Sal=salary*(1+yearlyIncrement);
y2Sal=y1Sal*(1+yearlyIncrement);
y3Sal=y2Sal*(1+yearlyIncrement);
y4Sal=y3Sal*(1+yearlyIncrement);
y5Sal=y4Sal*(1+yearlyIncrement);


y1Sav=y1Sal*yearlySavings;
y2Sav=y2Sal*yearlySavings;
y3Sav=y3Sal*yearlySavings;
y4Sav=y4Sal*yearlySavings;
y5Sav=y5Sal*yearlySavings;

print(f"1st Year Salary : ${y1Sal:.2f}")
print(f"1st Year Savings : ${y1Sal:.2f}")

print(f"2nd Year Salary : ${y2Sal:.2f}")
print(f"2nd Year Savings : ${y2Sal:.2f}")

print(f"3rd Year Salary : ${y3Sal:.2f}")
print(f"3rd Year Savings : ${y3Sal:.2f}")

print(f"4th Year Salary : ${y4Sal:.2f}")
print(f"4th Year Savings : ${y4Sal:.2f}")

print(f"5th Year Salary : ${y5Sal:.2f}")
print(f"5th Year Savings : ${y5Sav:.2f}")

print(f"5 Year Savings : ${(y1Sav+y2Sav+y3Sav+y4Sav+y5Sav):.2f}")


```
